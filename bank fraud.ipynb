```python
import numpy as np
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import roc_curve, average_precision_score

def prep(path):
    d = pd.read_csv(path).drop_duplicates()
    d = d.drop(columns=[c for c in ["x1", "x2"] if c in d.columns])
    for c in ["prev_address_months_count", "bank_months_count", "current_address_months_count",
              "session_length_in_minutes", "device_distinct_emails_8w"]:
        d[c + "_missing"] = (d[c] == -1).astype(int)
    d["balcon_neg"] = (d["intended_balcon_amount"] < 0).astype(int)
    d["source"] = (d["source"] == "TELEAPP").astype(int)
    d = pd.get_dummies(d, columns=["payment_type", "employment_status", "housing_status", "device_os"], drop_first=True)
    return d

drop2 = ["fraud_bool", "month", "device_fraud_count", "customer_age"]
labs = ["<=20", "30s", "40s", "50s", "60+"]
bins = [0, 20, 30, 40, 50, 100]

def line_at(y, s, target=0.05):
    f, _, th = roc_curve(y, s)
    return th[min(np.searchsorted(f, target), len(th) - 1)]

base = prep("Base.csv")
tr = base[base.month <= 1]; va = base[base.month == 2]
Xtr, ytr = tr.drop(columns=drop2), tr.fraud_bool
Xva, yva = va.drop(columns=drop2), va.fraud_bool

w = np.where(ytr == 1, (ytr == 0).sum() / (ytr == 1).sum(), 1.0)
m2 = HistGradientBoostingClassifier(max_iter=300, learning_rate=0.05, random_state=0)
m2.fit(Xtr, ytr, sample_weight=w)
pva = m2.predict_proba(Xva)[:, 1]
bva = pd.cut(va["customer_age"], bins, labels=labs).values
lines = {b: line_at(yva.values[bva == b], pva[bva == b]) for b in labs}
tg = line_at(yva.values, pva)
Xcols = Xtr.columns

rows = []
for name in ["Base", "Variant I", "Variant II", "Variant III", "Variant IV", "Variant V"]:
    v = prep(f"{name}.csv")
    v = v[v.month >= 3]
    X = v.drop(columns=drop2, errors="ignore").reindex(columns=Xcols, fill_value=0)
    y = v.fraud_bool.values
    s = m2.predict_proba(X)[:, 1]
    band = pd.cut(v["customer_age"], bins, labels=labs).values
    f, t, _ = roc_curve(y, s)
    flag_b = np.zeros(len(y), bool)
    for b in labs:
        flag_b[band == b] = s[band == b] >= lines[b]
    fp_global = [(s[(band == b) & (y == 0)] >= tg).mean() * 100 for b in labs]
    fp_band = [flag_b[(band == b) & (y == 0)].mean() * 100 for b in labs]
    rows.append({"data": name, "fraud_%": round(100 * y.mean(), 2),
                 "PR-AUC": round(average_precision_score(y, s), 3),
                 "TPR@5%FPR": round(np.interp(0.05, f, t), 3),
                 "gap_global": round(max(fp_global) - min(fp_global), 1),
                 "gap_perband": round(max(fp_band) - min(fp_band), 1)})
    print(rows[-1], flush=True)
print(pd.DataFrame(rows).to_string(index=False))
```
